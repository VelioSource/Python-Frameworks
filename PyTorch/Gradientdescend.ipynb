# Compute gradient yourself using numpy (every step manually)

Let's assume we want to use a model that implements linear regression.

So we have a formula of the form `f = w * x` 

```python
import numpy as np 
```

```python
# here : f = 2 * x
X = np.array([1, 2, 3, 4], dtype=np.float32)
Y = np.array([2, 4, 6, 8], dtype=np.float32)

w = 0.0
```

```python
# model output
def forward(x):
    return w * x
```

```python
# loss = MSE
def loss(y, y_pred):
    return ((y_pred - y)**2).mean()
```

```python
# Here we need to calculate the gradient of the loss-function ourselves.
# Loss = MSE = 1/N * (w*x - y)**2
# dLoss/dw = 1/N * 2 * (w*x - y) * x = 1/N * 2 * x * (w*x-y)
def gradient(x, y, y_pred):
    return np.dot(2*x, y_pred - y).mean()

print(f'Prediction before training: f(5) = {forward(5):.3f}')
```

```python
# Training
```

```python
learning_rate = 0.01
epochs = 20
```

```python
for epoch in range(epochs):
    # predict = forward pass
    y_pred = forward(X)

    # loss
    l = loss(Y, y_pred)
    
    # calculate gradients
    dw = gradient(X, Y, y_pred)

    # update weights
    w -= learning_rate * dw

    if epoch % 1 == 0:   # Look at all epochs
        print(f'epoch {epoch+1}: w = {w:.3f}, loss = {l:.8f}')
     
print(f'Prediction after training: f(5) = {forward(5):.3f}')
```

# Now, let's try with PyTorch

```python
# 1) Design model (input, output, forward pass with different layers)
# 2) Construct loss and optimizer
# 3) Training loop
#       - Forward = compute prediction and loss
#       - Backward = compute gradients
#       - Update weights



import torch
import torch.nn as nn
```

```python
# here : f = 2 * x

# 0) Training samples, watch the shape!
X = torch.tensor([[1], [2], [3], [4]], dtype=torch.float32)
Y = torch.tensor([[2], [4], [6], [8]], dtype=torch.float32)
```

```python
n_samples, n_features = X.shape

# 0) create a test sample
X_test = torch.tensor([5], dtype=torch.float32)
```

```python
# 1) Design Model, the model has to implement the forward pass!
# Here we can use a built-in model from PyTorch
input_size = n_features
output_size = n_features

model = nn.Linear(input_size, output_size)


print(f'Prediction before training: f(5) = {model(X_test).item():.3f}')
```

```python
# 2) Define loss and optimizer
l_rate = 0.05
n_iters = 500

loss = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=l_rate)
```

```python

# 3) Training loop
for epoch in range(n_iters):
    y_predicted = model(X)
    l = loss(Y, y_predicted)
    l.backward()
    optimizer.step()
    optimizer.zero_grad()

    if epoch % 10 == 0:
        [w, b] = model.parameters() # unpack parameters
        print('epoch ', epoch+1, ': w=', w[0][0].item(), ' loss = ', l)

print(f'Prediction after training: f(5) = {model(X_test).item():.3f}')
```
