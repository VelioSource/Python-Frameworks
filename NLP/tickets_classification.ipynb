## Data Processing

```python
data[data["Title"].isnull()]

```

```python
#Give a title for the ticket
data.loc[169, "Title"] = "Slow Network"
data.loc[186, "Title"] = "Data Privacy Concern"
```

```python
data[data["Description"].isnull()]
#data = data.drop(index=190)
data.drop(columns=data.columns[0], inplace=True)
```

```python
#What to do with the missing ticket? remove it
```

```python
#join the title with the ticket description
title_desc = data["Title"].astype(str) + ". " + data["Description"]
```

```python
data = pd.concat([data, title_desc], axis = 1)
data
```

```python
data.loc[[72]]
```

```python
#Drop possible columns 
data.drop(["Description", "Title"], axis=1, inplace=True)
```

```python
#rename again the column to: Description
data.rename(columns={data.columns[1]:"Description"},inplace=True)
data
```

```python
X = data['Description'] 
y = data['Class']
```

Split your data already to be sure that no data likeage will happens

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state = 4)
```

```python
train = pd.concat([X_train, y_train], axis =1)
train.to_csv('train_tickets.csv')
```

```python
test = pd.concat([X_test, y_test], axis =1)
test.to_csv('test_tickets.csv')
```
